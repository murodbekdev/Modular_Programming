# Pipelinega doir misollar 

```python
"""
[2, 5, 8, 10]

1. Har bir sonni 2 ga ko'paytirish
        ↓
2. 10 dan katta sonlarni qoldirish
        ↓
3. Yig'indisini hisoblash
"""

def multiply(list):
    new_list = []
    for each in list:
        new_list.append(each * 2)
    return new_list
print(multiply([2,4,5])) 

def greater_ten(list):
    new_list = []
    new_list2 = []
    for each in list:
        new_list.append(each * 2)
    for g in new_list:
        if g > 10:
            new_list2.append(g)
    return new_list2
print(greater_ten([2,4,5])) 

def summation(list):
    return sum(list)
print(summation([2,4,5])) 

# Pipeline 
pipeline = [multiply, greater_ten, summation]
list = [5, 6, 8, 9, 10]
for hip in pipeline:
    step = hip(list)
print(step)
```

```python
"""

1. 60 dan past ballarni olib tashlash
        ↓
2. Qolganlarini saralash
        ↓
3. O'rtacha qiymatini hisoblash


"""
def grades_sixty(grades):
    greater_than_sixty = []
    for grade in grades:
        if grade > 60:
            greater_than_sixty.append(grade)
    return greater_than_sixty
print(grades_sixty([60,67,67,89,90]))

def sorting_grades(grades):
    return sorted(grades) 
print(sorting_grades([98,78,67,99]))

def average(grades):
    return sum(grades) / len(grades)
print(average([98,78,67,99]))

# Pipeline 
grades_pipeline = [grades_sixty, sorting_grades, average]
grades = [8,98,77,87,78,67,99]
for grade in grades_pipeline:
    step = grade(grades)
print(step)
```

```python
"""
Input:
"  MURODJON@GMAIL.COM  "

        ↓

1. Bo'shliqlarni olib tashlash
        ↓
2. Lowercase qilish
        ↓
3. "@gmail.com" borligini tekshirish

"""
def remove_space(gmail):
    return gmail.replace(" ", "")
print(remove_space(" GAMIL@.COM "))

def to_lower(gmail):
    return gmail.lower()
print(to_lower(" GAMIL@.COM "))

def check_gmail(gmail):
    if "@gmail.com" in gmail:
        return gmail
    else:
        return "Invalid Email"
print(check_gmail(" GAMIL@.COM "))

# Pipeline Email 
pipeline_gmail = [remove_space, to_lower, check_gmail]
gmail = "mURodJOn@gmail.com "
for step in pipeline_gmail:
    gmail = step(gmail)
print(gmail)
```

```python
"""
Input:
"  MyPass123  "

        ↓

1. Bo'shliqlarni olib tashlash
        ↓
2. Parol uzunligini tekshirish
        ↓
3. Kuchli yoki kuchsizligini chiqarish

"""
def remove_space(password):
    return password.replace(" ", "")
def lengh_psw(password):
    return len(password)
def check_strong_psw(password):
    if password > 10:
        return "Strong Password"
    else:
        return "Not Strong Password"
# Pipelin for the Strong Password 
pipeline_psw = [remove_space,lengh_psw,check_strong_psw]
password = " asdasdsadsad9435345#@"
for step in pipeline_psw:
    password = step(password)
print(password)
```
